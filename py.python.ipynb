```python
 #sum of digits in a number
number= int(input("enter number:"))
total = 0 
while number > 0:
    digit = number % 10
    total = total + digit
    number = number // 10

print("sum of digits:", total)
```

```python
#reverse a number
number = int(input("enter number"))
reverse = 0

while number > 0:
    digit = number % 10
    number = number //10
    reverse = reverse * 10 + digit

print("reverse:", reverse) 
```

```python
#check if a number is prime
number = int(input("enter number:"))
count = 0
for i in range(1, number + 1):
    if number % i == 0:
        count = count + 1

if count == 2:
    print("prime number")
else:
    print("not a prime number")
    
  
```
