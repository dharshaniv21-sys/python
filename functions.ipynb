```python
#largest of three numbers
def largest(a,b,c):
    if a >=b and a>=c:
        return a
    elif b >= a and b >= c:
        return b
    else:
        return c
a = int(input("enter a:"))
b = int(input("enter b:"))
c = int(input("enter c:"))
print("largest=",largest(a,b,c))
```

```python
#grade calculation
def find_grade(marks):
    if marks >=90:
        return "A"
    elif marks >=80:
        return "B"
    elif marks >=70:
        return "C"
    elif marks >=60:
        return "D"
    else:
        return "FAIL"
marks = int(input("enter marks:"))
print("grade =", find_grade(marks))
```
